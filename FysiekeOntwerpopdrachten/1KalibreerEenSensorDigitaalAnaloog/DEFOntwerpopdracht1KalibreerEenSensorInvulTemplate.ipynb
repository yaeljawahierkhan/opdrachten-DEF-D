## *Opdracht 6*: Meten en fitten, Leerdoel

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

A_data = np.array([
    20, 32, 40, 48, 57,
    78, 82, 88, 92, 94, 98,
    111, 115, 117, 119, 131
])

W_data = np.array([
     47.84,
     95.69,
    143.54,
    192.55,
    240.40,
    432.90,
    480.75,
    529.76,
    577.60,
    625.45,
    673.30,
    998.20,
   1046.05,
   1095.06,
   1142.90,
   1452.60
])*9.81/1000
```

```python
def lineare_fit(x, a):
    return a * x**0.5

val, cov = curve_fit(lineare_fit, W_data, A_data)

x_test = np.arange(max(W_data) + 10)

plt.figure()
plt.plot(W_data, A_data, 'k.')
plt.errorbar(W_data, A_data, yerr=1,xerr=9.81/1000, fmt='k.', capsize=5)
plt.plot(x_test, lineare_fit(x_test, *val), 'r--')
plt.xlabel('Kracht [N]')
plt.ylabel('Waarde van arduino')
plt.show()
```
