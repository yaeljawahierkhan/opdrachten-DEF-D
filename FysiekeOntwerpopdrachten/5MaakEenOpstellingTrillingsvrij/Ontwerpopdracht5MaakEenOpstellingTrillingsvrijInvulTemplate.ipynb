## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
f_0 = 5/2
```

```python
def Amp(f):
    return np.abs(1/(1-(((2*np.pi*f)**2)/((f_0*np.pi*2)**2))))


test_freq = np.linspace(-10,2000,10000)

plt.figure()
plt.loglog(test_freq, Amp(test_freq))
plt.xlabel("f [Hz]")
plt.ylabel("|A_m/A_v| [-]")
plt.ylim(0.1,100)
plt.xlim(0,100)
plt.show()
```

```python
g = 9.81

delta_x = g/((f_0*2*np.pi)**2)

print("De delta_x waarbij je aan de eis voldoet is %.3f [m]" %delta_x)
```

## Opdracht 3: Plot de grafiek.

```python

def c(m,delta_x_func):
    return (m*g)/delta_x_func

m_test = np.linspace(0,3,100)

plt.figure()
plt.plot(m_test,c(m_test,delta_x),label="1*delta_x")
plt.plot(m_test,c(m_test,2*delta_x),label="2*delta_x")
plt.plot(m_test,c(m_test,5*delta_x),label="5*delta_x")
plt.plot(m_test,c(m_test,0.06),label="delta_x = 0.06")
plt.legend()
plt.xlabel("massa [kg]")
plt.ylabel("Veerconstante [N/m]")
plt.savefig("Verband veerconstante massa")
plt.show()
```
