De oppervlakte lading $\sigma(r)$ van een geleidende schijf met straal $R$ is niet uniform verdeeld over de schijf, maar wordt gegeven door (Problems 2.58 and 2.63 Griffiths 5th edition):
$$\sigma(r) = \frac{Q}{2 \pi R \sqrt{R^2 - r^2}}$$

1. Bereken de totale lading op de schijf met behulp van vraag 3a en 3b.

4. Plot in hetzelfde figuur de oppervlaktelading van een uniform geladen schijf met dezelfde lading $Q$.


````python
#3c.1:
Q = 2 * sigma_center * np.pi * disk_radius**2
print("Totale lading op de schijf:")
print(f"Q_tot = {Q:.3e} C\n")

#3c.2:
R = disk_radius
sigma_uniform = Q / (np.pi * R**2)

print("Oppervlaktelading in het centrum:")
print(f"Geleidende schijf : σ(0) = {sigma_center:.3e} C/m²")
print(f"Uniforme schijf   : σ    = {sigma_uniform:.3e} C/m²")
#De verhouding geleidende schijf (r=0) en uniforme schijf is 0.5, wat logisch volgt uit de formules.

#3c.3:
r = np.linspace(0, 0.999 * R, 500)

sigma_conducting = Q / (2 * np.pi * R * np.sqrt(R**2 - r**2))
sigma_uniform_r = Q/(np.pi * R**2) + np.zeros(len(r))
plt.figure()
plt.plot(r, sigma_conducting, label="Geleidende schijf")
plt.plot(r, sigma_uniform_r, "--", label="Uniform geladen schijf (r=0)")
plt.xlabel("r (m)")
plt.ylabel("σ (C/m²)")
plt.title("Oppervlaktelading van een schijf")
plt.legend()
plt.show()
````
